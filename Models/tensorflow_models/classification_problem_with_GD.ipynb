```python
def initilizer_parameters(layer_dims):
    np.random.seed(3)
    parameters = {}
    for l in range(1, len(layer_dims)):
        parameters["W" + str(l)] = np.random.randn(layer_dims[l-1], layer_dims[l]) * 0.01
        parameters["b" + str(l)] = np.zeros((layer_dims[l], 1))
    return parameters
```

```python
initilizer_parameters([2,2,1])
```

```python
def sigmoid(z):
    A = 1/(1+np.exp(-z))
    return A
```

```python
def linear_forward(A_prev, W, b): 
    z = np.dot(W.T, A_prev) +b 

    A = sigmoid(z)
    return A
```

```python
def L_layer_forwad(X, parameters):
    A = X
    L = len(parameters)//2

    for i in range(1, L+1):
        A_prev = A
        W1 = parameters['W' + str(i)]
        b1 = parameters['b' + str(i)]
        
        print("A"+ str(i-1) + ":", A_prev)
        print("W"+ str(i) + ":", W1)
        print("b"+ str(i) + ":", b1)
        print("--"* 20)
        
        A = linear_forward(A_prev, W1, b1)
        print("A"+ str(i) + ":", A)
        print("**"* 20)
    return A,A_prev
```

```python
def update_parameters(parameters, y, y_hat, A1,X):
    parameters['W2'][0][0] = parameters['W2'][0][0] + (0.01 * (y-y_hat)*A1[0][0])
    parameters['W2'][1][0] = parameters['W2'][1][0] + (0.01 * (y-y_hat)*A1[1][0])
    parameters['b2'][0][0] = parameters['W2'][1][0] + (0.01 * (y-y_hat))

    parameters['W1'][0][0] = parameters['W1'][0][0] + (0.01 * (y-y_hat)*parameters['W2'][0][0]*(1-A1[0][0])*X[0][0])
    parameters['W1'][0][1] = parameters['W1'][0][1] + (0.01 * (y-y_hat)*parameters['W2'][0][0]*(1-A1[0][0])*X[1][0])
    parameters['b1'][0][0] = parameters['b1'][0][0] + (0.01 * (y-y_hat)*parameters['W2'][0][0]*(1-A1[0][0]))                                          

    parameters['W1'][1][0] = parameters['W1'][1][0] + (0.01 * (y-y_hat)*parameters['W2'][1][0]*(1-A1[0][0])*X[0][0])
    parameters['W1'][1][1] = parameters['W1'][1][1] + (0.01 * (y-y_hat)*parameters['W2'][1][0]*(1-A1[0][0])*X[1][0])
    parameters['b1'][1][0] = parameters['b1'][1][0] + (0.01 * (y-y_hat)*parameters['W2'][1][0]*(1-A1[0][0]))                                          
```

```python
X = df[['cgpa', 'profile_score']].values[0].reshape(2,1)
y =df[['placed']].values[0][0]

#Parameter_initializtion
parameters = initilizer_parameters([2,2,1])

y_hat, A1 = L_layer_forwad(X, parameters)
```

```python
y_hat = y_hat[0][0]
```

```python
update_parameters(parameters, y, y_hat, A1,X)
```

```python
print("Loss for this student :" ,-y*np.log(y_hat)-(1-y)*np.log(1-y_hat))
parameters
```

```python
parameter = initilizer_parameters([2,2,1])
epchos = 5

for i in range(epchos):
    Loss = []

    for j in range(df.shape[0]):
        x = df[['cgpa', 'profile_score']].values[j].reshape(2,1)
        y =df[['placed']].values[j][0]

        y_hat, A1 = L_layer_forwad(X, parameters)
        y_hat = y_hat[0][0]

        update_parameters(parameter, y, y_hat, A1,x)

        Loss.append(-y*np.log(y_hat)-(1-y)*np.log(1-y_hat))
    print('Epchos - ',i+1, 'Loss - ', np.array(Loss).mean())
parameter
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(2, activation = 'sigmoid', input_dim = 2))
model.add(Dense(1, activation = 'sigmoid'))
```

```python
new_weights = [np.array([[0.1, 0.1],[0.1, 0.1]], dtype=np.float32),
np.array([0.,0.], dtype=np.float32),np.array([[0.1],[0.1]], dtype=np.float32),
np.array([0.], dtype=np.float32)]
```

```python
model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
optimizer = keras.optimizers.Adam(learning_rate = 0.001)
model.compile(loss = 'binary_crossentropy',optimizer = optimizer )
```

```python
model.fit(df.iloc[:,0:-1].values, df['placed'].values, epochs = 75, verbose = 1, batch_size =1)
```
